```python
print(A)
```

```python
print(la.det(A))
```

```python
A_inv= la.inv(A)
print(A_inv)
```

```python
print(A@A_inv)
```

```python
a= np.array([[1,1,1]],)
b= np.array([[1,0,-1]],)
c= np.array([[1,2,-1]],)
d= np.array([[1,2,0]],)
e= np.array([[0,1,1]],)
f= np.array([[1,0,2]],)
g= np.array([[2,1,3]],)
```

```python
np.matmul(a.T,b) # 	NumPy's function for matrix multiplication, np.matmul() and a.T is a transpose of matrix a
```

```python
print(np.shape(a))
print(np.shape(b))
```

```python
h= np.vstack((d,e)) # Stacks the array vertically, d becomes the first row and e the second row
i= np.hstack((f.T,g.T))  # Stacks the array horizontally

print(h)
print()
print(i)
```

```python
j= h@i #@ is the matrix product of h and i, just like np.matmul(h, i) or np.dot(h, i)
print(j)
```
